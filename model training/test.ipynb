```python
import yaml

# Path to the data.yaml file
data_yaml_path = "/content/Pothole-2/data.yaml"

# Read the existing data.yaml file
with open(data_yaml_path, 'r') as f:
    data = yaml.safe_load(f)

# Correct the relative paths
data['train'] = data['train'].replace('../', './')
data['val'] = data['val'].replace('../', './')
data['test'] = data['test'].replace('../', './')

# Write the updated data.yaml back to the file
with open(data_yaml_path, 'w') as f:
    yaml.dump(data, f)

print(f"Updated {data_yaml_path} with corrected relative paths.")
print("New content of data.yaml:")
with open(data_yaml_path, 'r') as f:
    print(f.read())
```
